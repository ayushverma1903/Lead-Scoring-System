```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/Lead Scoring.csv")
print("Original Shape:", df.shape)
```

```python
df.drop(columns=['Prospect ID', 'Lead Number', 'Magazine'], inplace=True)
print("Shape after dropping useless columns:", df.shape)
```

```python
# Replace 'Select' with NaN across the dataset
df.replace('Select', np.nan, inplace=True)

# Re-check missing values after this fix
missing = df.isnull().sum().sort_values(ascending=False)
missing_percent = (missing / len(df)) * 100
missing_df = pd.DataFrame({"Missing Count": missing, "Missing %": missing_percent})
missing_df[missing_df["Missing Count"] > 0]
```

```python
drop_cols = [
    'How did you hear about X Education',
    'Lead Profile',
    'Asymmetrique Activity Index',
    'Asymmetrique Profile Index',
    'Asymmetrique Activity Score',
    'Asymmetrique Profile Score'
]

df.drop(columns=drop_cols, inplace=True)
print("Shape after dropping high-missing columns:", df.shape)
```

```python
categorical_fill_unknown = [
    'Lead Quality', 'City', 'Specialization', 'Tags',
    'What matters most to you in choosing a course',
    'What is your current occupation', 'Country'
]

for col in categorical_fill_unknown:
    df[col] = df[col].fillna('Unknown')

print("Missing values remaining in these columns:")
print(df[categorical_fill_unknown].isnull().sum())
```

```python
df['TotalVisits'] = df['TotalVisits'].fillna(df['TotalVisits'].median())
df['Page Views Per Visit'] = df['Page Views Per Visit'].fillna(df['Page Views Per Visit'].median())
```

```python
print("Total missing values remaining:", df.isnull().sum().sum())
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
df['Lead Source'] = df['Lead Source'].fillna(df['Lead Source'].mode()[0])
df['Last Activity'] = df['Last Activity'].fillna(df['Last Activity'].mode()[0])

# Final check
print("Total missing values remaining:", df.isnull().sum().sum())
```

```python
df.to_csv("../data/processed/cleaned_leads.csv", index=False)
print("✅ Cleaned dataset saved to data/processed/cleaned_leads.csv")
print("Final shape:", df.shape)
```
